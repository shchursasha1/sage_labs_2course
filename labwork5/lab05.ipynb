## Завдання 2 

Розгляньте групу матриць за множенням: 

$$G = \left\langle \left(\begin{array}{rr}
1 & -1 \\
-1 & 2
\end{array}\right), \left(\begin{array}{rr}
3 & 2 \\
2 & 1
\end{array}\right) \right\rangle$$

Дайте відповідь на наступні питання: 

1. _Чи є група G скінченною?_
2. _Чи є група G комутативною?_
3. _Який вигляд мають елементи групи G?_
4. _Які порядки мають елементи групи G?_ 
5. _Чому ізоморфна група G?_


__Відповіді__: 
1. Група не скінченна. 
2. Група комутативна. 
3.  Група \( G \) складається з матриць, породжених двома даними матрицями:
   \[
   A = \begin{pmatrix} 1 & -1 \\ -1 & 2 \end{pmatrix} B = \begin{pmatrix} 3 & 2 \\ 2 & 1 \end{pmatrix}.
   \]
   Елементи групи \( G \) будуть всіма можливими добутками цих матриць і їх обернених (оскільки група повинна містити обернені елементи для кожного елемента, а також одиничний елемент). Тому елементи \( G \) – це всі можливі лінійні комбінації цих двох матриць з урахуванням добутків та обернених. 
4. Елементи мають нескінченні порядки. 
5.   Оскільки група \( G \) є комутативною і має нескінченні порядки елементів, можна припустити, що вона ізоморфна прямій сумі нескінченних циклічних груп.

```python
var("k m")

A = Matrix([[1, -1],
            [-1, 2]])

B = Matrix([[3, 2],
            [2, 1]])

print("Порядки A:")
for power in range(1, 5):
    print(f"A^{power} =\n{A^power}\n")

print("Порядки B:")
for power in range(1, 5):
    print(f"B^{power} =\n{B^power}\n")

print("Порядки (A * B)^n:")
for power in range(1, 5):
    print(f"(A * B)^{power} =\n{(A * B)^power}\n")
    
    
print("Вигляд елементів A^k * B^m")   
general_form = A^k * B^m
print(general_form)

if general_form == (B^m * A^k):
    print("Група є комутативною")
else:
    print("Група не є комутативною")
```
