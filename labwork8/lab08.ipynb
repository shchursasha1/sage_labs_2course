__Знайдіть к-ть різних розфарбувань кубика Рубика 2х2х2, використовуючи 10 кольорів.__


![rubik cube](../docs/labs_assets/rubik_cube.jpg)

Два розфарбування називатимуться однаковими, якщо існує послідовність рухів кубика рубика, яка одне розфарбування переводить в інше. 

Не обовʼязково використовувати всі 10 кольорів, тобто допустимо пофарбувати весь кубик в один колір. 

___Приклад___: існує 183 різних розфарбувань, використовуючи 2 кольори. 

```python
SymGroup = SymmetricGroup(24)

gen1 = SymGroup('(1, 2, 4, 3)(5, 6, 8, 7)(9, 10, 12, 11)(13, 14, 16, 15)(17, 18, 20, 19)(21, 22, 24, 23)')
gen2 = SymGroup('(1, 5, 9, 13, 17, 21)(2, 6, 10, 14, 18, 22)(3, 7, 11, 15, 19, 23)(4, 8, 12, 16, 20, 24)')
gen3 = SymGroup('(1, 3)(2, 4)(5, 7)(6, 8)(9, 11)(10, 12)(13, 15)(14, 16)(17, 19)(18, 20)(21, 23)(22, 24)')
gen4 = SymGroup('(1, 9)(2, 10)(3, 11)(4, 12)(5, 13)(6, 14)(7, 15)(8, 16)(17, 21)(18, 22)(19, 23)(20, 24)')
gen5 = SymGroup('(5, 9)(6, 10)(7, 11)(8, 12)(13, 17)(14, 18)(15, 19)(16, 20)(21, 23)(22, 24)')
gen6 = SymGroup('(13, 21)(14, 22)(15, 23)(16, 24)')

CubeSymGroup = SymGroup.subgroup([gen1, gen2, gen3, gen4, gen5, gen6])

print("Розмір групи симетрій:", CubeSymGroup.order())

colors_count = 10

def count_colorings(permutation, colors):
    cycle_structure = permutation.cycle_type()
    return colors**sum(cycle_structure)

total_colorings = sum(count_colorings(elem, colors_count) for elem in CubeSymGroup) / CubeSymGroup.order()

print("Кількість унікальних розфарбувань:", total_colorings)
```
