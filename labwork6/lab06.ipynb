## Завдання 1

В групі рухів правильного восьмикутника задайте циклічну підгрупу порядку 4. Побудуйте для неї ліві і праві класи суміжності та перевірте чи буде ця група нормальною. 

```python
import time

def calculate_left_classes(group, subgroup):
    left_classes = []
    for elem in group:
        new_class = [elem * sub_elem for sub_elem in subgroup]
        new_class_sorted = sorted(new_class)
        if new_class_sorted not in left_classes:
            left_classes.append(new_class_sorted)
    return left_classes

def calculate_right_classes(group, subgroup):
    right_classes = []
    for elem in group:
        new_class = [sub_elem * elem for sub_elem in subgroup]
        new_class_sorted = sorted(new_class)
        if new_class_sorted not in right_classes:
            right_classes.append(new_class_sorted)
    return right_classes

D = DihedralGroup(8)
elements_of_order_4 = [el for el in D if el.order() == 4]
G = D.subgroup(elements_of_order_4)
print("Subgroup G:", G)

left_classes = calculate_left_classes(D, G)
right_classes = calculate_right_classes(D, G)

print("\nLeft classes:")
for left_class in left_classes:
    print(left_class)

print("\nRight classes:")
for right_class in right_classes:
    print(right_class)

print()
if left_classes == right_classes:
    print("Left and right classes are equal, so this subgroup is normal.")
print(f'Is subgroup normal: {G.is_normal()}')
```

Спрогнозуйте скільки нормальних підгруп має група $D_{470448}$ __без виводу всіх за допомогою Sage__. _Опишіть_ всі нормальні підгрупи групи $D_{470448}$.

Гіпотеза:

1. Загальна структура:
       Якщо порядок групи $D_n$ парний, то всі дільники порядку групи є можливими порядками її нормальних підгруп.
       Якщо порядок групи парний, то існує \textbf{рівно три нормальні підгрупи} з порядком $\frac{n}{2}$, де $n$ — порядок групи.

2. Застосування до $D_{470448}$:
        Оскільки порядок групи $D_{470448}$ дорівнює $940896$ (тобто $2 \times 470448$), всі дільники числа $940896$ будуть порядками нормальних підгруп.
        Серед цих підгруп буде рівно три підгрупи порядку $470448$.

Висновок

Таким чином, діедральна група $D_{470448}$ має нормальні підгрупи з порядками, що дорівнюють усім дільникам числа $940896$, і серед них буде три нормальні підгрупи з порядком $470448$.


```python
n_values = range(3, 101)
normal_subgroup_orders = {}

for n in n_values:
    D_n = DihedralGroup(n)
    normal_subgroups = D_n.normal_subgroups()
    orders = [subgroup.order() for subgroup in normal_subgroups]
    normal_subgroup_orders[n] = sorted(orders)

for n, orders in normal_subgroup_orders.items():
    print(f"D_{n}: порядки нормальних підгруп: {orders}")
```
